## 13. Summary and limitations
- Clean data: no missing values, no duplicates; the hidden BOM in the first column header is removed in `load_data`.
- Area (r = 0.86) and Location are the main price drivers; Type and Age add a smaller but real effect.
- Floor, Distance, Furnishing, Parking and all time features show no meaningful effect on price.
- Bedrooms and Bathrooms are collinear with Area; use `Bath_Bed_Ratio` or drop them.
- The date and time columns look randomly assigned: no trend, seasonality or intervention, one observation per property.
- Some patterns are unusual (houses on upper floors, no distance effect), so the data may be synthetic. This is an inference, not a proven fact.
- Synthetic control and a real pricing-lift estimate still need market data with repeated observations over time.
